# 4. 関数と関係

## 4.1. 写像

- 写像：集合$S$のすべての元に対して、集合$T$の元がただ一つ対応する規則（$f: S \to T$）があるとき、その規則$f$を$S$から$T$への写像という。また、$S$を定義域（始域）、$T$を値域（終域、像）という。
    - すべての$s \in S$に対応する$t \in T$が存在する必要がある
    - $s \in S$に対して複数の$t \in T$は対応しない

### 全射・単射・全単射

- $f: S \to T$において、$\forall t \in T$に対して、$\exists s \in S$が存在するとき、全射という。
- $f: S \to T$において、$a, b \in S$に対して、$a \neq b$の時、$f(a) \neq f(b)$ならば、単射という。
- 全射かつ単射の時全単射という。

## 4.2. 二項関係

### 直積集合

- 直積集合
    - 集合$A$、$B$の元$a \in A$、$b \in B$の対$(a,b)$の集合
    - $A \times B = \{(a,b) | a \in A, b \in B\}$

```python
A = {i for i in range(5)}
B = {'a', 'b', 'c'}

for a in A:
    for b in B:
        print((a,b))
```

## 二項関係

- ２つの集合の元を組み合わせる対応規則を二項関係という。
- 二項関係は直積集合の部分集合である。$R \subseteq A \times B$
- $(x,y) \in R$の時、$xRy$と表す。

### 対称と反対称

- 対称律：$xRy \Rightarrow yRx$（逆の関係も成り立つ）
- 反対称律：$xRy, yRx \Rightarrow x = y$（一方の関係が成り立つならば、その逆の関係は成り立たない）

### 関係の濃度

- 直積集合の濃度$|A \times B| = |A||B|$
- 直積集合の冪集合の濃度$|\mathcal{P}(A \times B)| = 2 ^ {|A||B|}$

### 合成

- $R: A \to B$、$S: B \to C$がある時、その合成を $S \circ R = \{(a, c) \in A \times C | \exists b : aRb \wedge bRc\}$と表す。

## 4.3. 同値関係と同値類

$S$上の二項関係で以下を満たすものを同値関係という

- 反射律：$\forall s \in S, sRs$
- 対称律：$\forall a,b \in S, aRb \Rightarrow bRa$
- 推移律：$\forall a,b,c \in S, aRb, bRc \Rightarrow aRc$

集合の中で同じとみなしていいもの（同じとみなしていい関係にあるもの）を定義できる

同値関係にあるSの互いに素な部分集合を同値類という。

→集合の中で、「同じ」と見做せるものを同値関係でグルーピングする、同値関係はグルーピングのための道具

$S$上の同値関係$R$と$s \in S$の元の関係を

$\bar{s} = \{x \in S | xRs\}$

と表す。

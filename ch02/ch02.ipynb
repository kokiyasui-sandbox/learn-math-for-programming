## 2.1. 集合について

- 集合
    - 物の集まり$S=\{1,2,3\}$のように表し、その個数$|S|=3$を濃度と呼ぶ
    - 例えば、20以下の偶数の集合は内包表現を用いて次のように表せる
        - $A = \{x | x \in \mathbb{N}, x < 20, x \mod 2 = 0\}$
    - $\emptyset$(空集合)は要素の存在しない集合
      

### 集合演算
|演算|定義|
|---|---|
|和集合|$A \cup B = \{x | x \in A \vee x \in B\}$|
|積集合|$A \cap B = \{x | x \in A \wedge x \in B \}$|
|差集合|$A - B = \{x | x \in A \wedge x \notin B\}$|
|対称差集合|$A \bigtriangleup B = \{x | (x \in A \wedge x \notin B) \wedge (x \notin A \wedge x \in B)\}$|

```python
# 集合の作り方
x = [1, 1, 2, 3, 4, 4, 5]
y = list(set(x))

print(y)
```

```python
A = {1, 2, 3, 4, 5, 6}
B = {0, 2, 4, 6, 8}

print(A|B)
print(A&B)
print(A-B)
print(A^B)
```

### 部分集合

- 集合$B$の元が集合$A$の元に含まれるとき、$B$を$A$の部分集合という。($B \subset A$)
- $B$と$A$が等しくない時、$B$を$A$の真部分集合という。同じ時、$B \subseteq A$とかく。

### 冪集合

- ある集合$A$のすべての部分集合の集合から作られた集合を冪集合という。
    - $|\mathcal{P}(A)| = 2 ^ {|A|}$

### 無限集合の濃度

- $\mathbb{N}$は無限可算集合（１、２と番号を振ることができる）
- その濃度を$|\mathbb{N}| = \aleph_0$とする
- $\mathbb{W}$、$\mathbb{Z}$、$\mathbb{Q}$も無限可算集合でそれらの濃度も$|\mathbb{N}|=|\mathbb{W}|=|\mathbb{Z}|=|\mathbb{Q}|=\aleph_0$
- カントールの対角線論法により、$|\mathbb{R}| > |\mathbb{N}|$である

# 1.コンピュータと数値

## 1.1.数と基数

```python
# binary
print(0b0101)
print(bin(5))
print(f"{5:b}")
```

```python
# octal
print(0o12)
print(oct(10))
print(f"{10:o}")
```

```python
# hex
print(0x14)
print(hex(20))
print(f"{20:x}")
```

```python
# convert base
def convertBase(n: int, b: int) -> str:
    ASCII_CODE_ZERO = 48
    ASCII_CODE_A    = 65

    def convertDigit(r: int) -> str:
        ascii_code : int = ASCII_CODE_ZERO + r if (r < 10) else ASCII_CODE_A + r - 10
        return chr(ascii_code)

    ret: str = ''

    while (n != 0):
        r = n % b
        n //= b
        ret = ret + convertDigit(r)

    return "".join(reversed(ret))

print(convertBase(10, 2))
```
